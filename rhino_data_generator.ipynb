# 🦏 Rhino Conservation Dataset Generator

Welcome to the **Rhino Conservation Dataset Generator Notebook**!  
This notebook simulates a dataset for analyzing conservation efforts for rhinos.  
Run all cells to generate the `rhino_data.csv` file for use in the Streamlit app. 🚀


````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set seed for reproducibility 🔒
np.random.seed(42)

# Simulate data
n = 500
zones = [f"Zone {i}" for i in range(1, 11)]
data = {
    "Zone": np.random.choice(zones, n),
    "RhinoCount": np.random.poisson(lam=5, size=n),
    "PoachingIncidents": np.random.binomial(n=1, p=0.2, size=n),
    "RangerPatrolFreq": np.random.normal(loc=7, scale=2, size=n).clip(min=1),
    "CameraSurveillance": np.random.choice([0, 1], size=n, p=[0.4, 0.6]),
    "WaterSources": np.random.randint(1, 5, size=n),
    "TouristActivity": np.random.normal(loc=50, scale=15, size=n).astype(int).clip(min=0)
}

df = pd.DataFrame(data)
df.columns = df.columns.str.strip()  # Remove extra spaces if any

# Save dataset
df.to_csv("rhino_data.csv", index=False)
print("✅ Dataset created successfully!")
````

````python

# Visualizations
# 1. Distribution of Poaching Incidents
plt.figure(figsize=(8, 5))
sns.countplot(x="PoachingIncidents", hue="PoachingIncidents", data=df, palette="Set2", legend=False)
plt.title("Distribution of Poaching Incidents")
plt.xlabel("Number of Incidents")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 2. Ranger Patrol Frequency vs Poaching Incidents
plt.figure(figsize=(8, 5))
sns.boxplot(x="PoachingIncidents", y="RangerPatrolFreq", data=df, palette="coolwarm")
plt.title("Ranger Patrol Frequency vs Poaching Incidents")
plt.xlabel("Poaching Incidents")
plt.ylabel("Patrol Frequency")
plt.tight_layout()
plt.show()

# 3. Human Activity vs Poaching Incidents by Zone
plt.figure(figsize=(8, 5))
sns.scatterplot(x="TouristActivity", y="PoachingIncidents", data=df, hue="Zone", palette="viridis")
plt.title("Tourist Activity vs Poaching Incidents by Zone")
plt.xlabel("Tourist Activity")
plt.ylabel("Poaching Incidents")
plt.tight_layout()
plt.show()

# 4. Correlation Heatmap
plt.figure(figsize=(8, 6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
````
